```python
import pandas as pd

import requests
import time
```

```python
# Gathering data from Steam API
games = {
    "borderlands_4": 1285190,
    "helldivers_2": 553850,
    "monster_hunter_wilds": 2246340
}

MAX_NUM_REVIEWS = 10_000
borderlands_4_df = None
helldivers_2 = None
monster_hunter_wilds = None

for game_name, game_id in games.items():
    print(f"\n🚀 STARTING DATA COLLECTION FOR: {game_name.upper()}...")

    all_reviews = [] # data['reviews'] is a list w/ dictionaries where each dict is a user's review & stats
    cursor = "*"
    
    url = f"https://store.steampowered.com/appreviews/{game_id}"

    while len(all_reviews) < MAX_NUM_REVIEWS:
        params = {
            "json": 1,
            "cursor": cursor,
            "num_per_page": 100,
            "filter": "recent",
            "language": "english"
        }

        response = requests.get(url, params)
        data = response.json()

        all_reviews.extend(data['reviews'])
        cursor = data['cursor'] # grab the next most recent batch of reviews
        
        print(f"Rows grabbed so far: {len(all_reviews)}")
        time.sleep(1) # to respect Steam

    # Creating dataframes from the API data
    df_reviews = pd.DataFrame(all_reviews)

    # Create a local csv file for my laptop
    df_reviews.to_csv(f"steam_reviews_{game_name}.csv", index=False)

    # Keeping data locally in Jupyter
    if game_name == 'borderlands_4':
        borderlands_4_df = df_reviews
    elif game_name == 'helldivers_2':
        helldivers_2 = df_reviews
    else:
        monster_hunter_wilds = df_reviews
```
